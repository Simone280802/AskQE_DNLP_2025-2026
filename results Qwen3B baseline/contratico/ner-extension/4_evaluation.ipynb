## 3. Results Visualization

```python
import json
import pandas as pd

# ── Load String Comparison results ──
with open(os.path.join(STRING_OUT, 'string_comparison_results.json'), 'r') as f:
    sc_results = json.load(f)

print('=== STRING COMPARISON — By Language ===')
df_lang = pd.DataFrame(sc_results['by_language']).T
display(df_lang)

print('\n=== STRING COMPARISON — By Perturbation ===')
df_pert = pd.DataFrame(sc_results['by_perturbation']).T
display(df_pert)

print('\n=== STRING COMPARISON — By Entity Type ===')
df_et = pd.DataFrame(sc_results['by_entity_type']).T.sort_values('count', ascending=False)
display(df_et)
```

```python
# ── Load SBERT results ──
with open(os.path.join(SBERT_OUT, 'sbert_results.json'), 'r') as f:
    sbert_results = json.load(f)

print('=== SBERT — By Language ===')
df_sbert_lang = pd.DataFrame(sbert_results['by_language']).T
display(df_sbert_lang)

print('\n=== SBERT — By Perturbation ===')
df_sbert_pert = pd.DataFrame(sbert_results['by_perturbation']).T
display(df_sbert_pert)

print('\n=== SBERT — By Entity Type ===')
df_sbert_et = pd.DataFrame(sbert_results['by_entity_type']).T.sort_values('count', ascending=False)
display(df_sbert_et)
```
