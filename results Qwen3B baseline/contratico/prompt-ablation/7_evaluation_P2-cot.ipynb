```python
import os
import sys
import subprocess

IN_COLAB = 'google.colab' in sys.modules
IN_KAGGLE = os.path.exists('/kaggle')

if IN_KAGGLE:
    PROJECT_ROOT = '/kaggle/working/askqe'
    if not os.path.exists(PROJECT_ROOT):
        subprocess.run(['git', 'clone',
                        'https://github.com/Simone280802/AskQE_DNLP_2025-2026.git',
                        PROJECT_ROOT], check=True)
elif IN_COLAB:
    PROJECT_ROOT = '/content/askqe'
    if not os.path.exists(PROJECT_ROOT):
        subprocess.run(['git', 'clone',
                        'https://github.com/Simone280802/AskQE_DNLP_2025-2026.git',
                        PROJECT_ROOT], check=True)
else:
    PROJECT_ROOT = os.getcwd()

print(f'Project root: {PROJECT_ROOT}')

# ─── Paths ───
PROMPT_ABLATION_DIR = f"{PROJECT_ROOT}/results Qwen3B baseline/contratico/prompt-ablation"
if IN_KAGGLE or IN_COLAB:
     PROMPT_ABLATION_DIR = "/kaggle/working/askqe/results Qwen3B baseline/contratico/prompt-ablation" if IN_KAGGLE else "/content/prompt-ablation"

CODE_DIR = f"{PROJECT_ROOT}/results Qwen3B baseline/contratico/prompt-ablation/code"

STRATEGIES = ["P2-cot"]

print(f"Evaluation Directory: {PROMPT_ABLATION_DIR}")
print(f"Code: {CODE_DIR}")
```

```python
import subprocess
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', 
                'sentence-transformers', 'nltk', 'pandas', 'matplotlib', 'seaborn', 'sacrebleu'], check=True)
print('Dependencies installed!')
```

```python
for strategy in STRATEGIES:
    print(f"\n{'=' * 60}")
    print(f"EVALUATION: {strategy}")
    print(f"{'=' * 60}")
    
    result = subprocess.run(
        [
            sys.executable, f"{CODE_DIR}/evaluation_contratico.py",
            "--prompt_ablation_dir", PROMPT_ABLATION_DIR,
            "--strategy", strategy,
        ],
        capture_output=True, text=True
    )
    print(result.stdout)
    if result.returncode != 0:
        print(f"ERROR: {result.stderr}")
```
