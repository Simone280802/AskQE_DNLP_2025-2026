# Prompt Ablation - contraTICO
## 7) Evaluation

Compute F1, EM, BLEU, chrF between Source QA and BT QA answers for all 3 strategies.

```python
import os
import subprocess

# ─── Paths (Kaggle) ───
PROJECT_ROOT = "/kaggle/input/askqe-dnlp"
PROMPT_ABLATION_DIR = f"/kaggle/working/prompt-ablation"
CODE_DIR = f"{PROJECT_ROOT}/results Qwen3B baseline/contratico/prompt-ablation/code"

STRATEGIES = ["P1-fewshot", "P2-cot", "P3-concise"]

print(f"Evaluation Directory: {PROMPT_ABLATION_DIR}")
print(f"Code: {CODE_DIR}")
```

```python
for strategy in STRATEGIES:
    print(f"\n{'=' * 60}")
    print(f"EVALUATION: {strategy}")
    print(f"{'=' * 60}")
    
    result = subprocess.run(
        [
            "python", f"{CODE_DIR}/evaluation_contratico.py",
            "--prompt_ablation_dir", PROMPT_ABLATION_DIR,
            "--strategy", strategy,
        ],
        capture_output=True, text=True
    )
    print(result.stdout)
    if result.returncode != 0:
        print(f"ERROR: {result.stderr}")
```
