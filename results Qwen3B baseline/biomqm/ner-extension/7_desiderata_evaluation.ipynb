# NER Extension — QG Desiderata Evaluation (biomqm)

Evaluate the quality of questions generated by the entity-aware QG pipeline.

**Metrics computed:**
1. Average number of questions per entry
2. Duplicate question detection
3. Diversity (SBERT cosine similarity + BERTScore)
4. Answerability (Longformer)
5. Readability (Flesch Reading Ease)

```python
import os
import sys
import subprocess

IN_COLAB = 'google.colab' in sys.modules
IN_KAGGLE = os.path.exists('/kaggle')

if IN_KAGGLE:
    PROJECT_ROOT = '/kaggle/working/askqe'
    if not os.path.exists(PROJECT_ROOT):
        subprocess.run(['git', 'clone',
                        'https://github.com/Simone280802/AskQE_DNLP_2025-2026.git',
                        PROJECT_ROOT], check=True)
elif IN_COLAB:
    PROJECT_ROOT = '/content/askqe'
    if not os.path.exists(PROJECT_ROOT):
        subprocess.run(['git', 'clone',
                        'https://github.com/Simone280802/AskQE_DNLP_2025-2026.git',
                        PROJECT_ROOT], check=True)
else:
    PROJECT_ROOT = os.getcwd()

print(f'Project root: {PROJECT_ROOT}')

NER_EXT_DIR = os.path.join(PROJECT_ROOT, 'results Qwen3B baseline', 'biomqm', 'ner-extension')
DESIDERATA_DIR = os.path.join(NER_EXT_DIR, 'evaluation', 'desiderata')

print(f'NER Extension: {NER_EXT_DIR}')
print(f'Desiderata scripts: {DESIDERATA_DIR}')
```
