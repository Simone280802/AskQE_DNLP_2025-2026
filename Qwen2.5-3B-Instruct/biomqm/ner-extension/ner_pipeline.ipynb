# BioMQM — NER Extension Pipeline

Full entity-aware pipeline:
1. **NER Extraction** — Extract biomedical entities (`ner_extraction.py`)
2. **QG** — Entity-aware question generation (`qg_entity_aware.py`)
3. **QA Source** — Answer on English source (`qa_entity.py --mode source`)
4. **QA BT** — Answer on back-translations (`qa_entity.py --mode bt`) × 5 languages
5. **Mapping + Evaluation** — String comparison and SBERT

| Parameter | Values |
|-----------|--------|
| NER model | d4data/biomedical-ner-all |
| LLM | Qwen/Qwen2.5-3B-Instruct |
| Languages | de, es, fr, ru, zh-CN |

## 1. Configuration

````python
NER_DIR    = f'{PROJECT_ROOT}/Qwen2.5-3B-Instruct/biomqm/ner-extension'
CODE_DIR   = f'{NER_DIR}/code'
DATA_DIR   = f'{PROJECT_ROOT}/biomqm'
LANGUAGES  = ['de', 'es', 'fr', 'ru', 'zh-CN']

# Create output directories
for d in ['results/ner', 'results/qg', 'results/qa/source', 'results/qa/bt',
          'results/mapping', 'results/sbert', 'results/string_comparison']:
    os.makedirs(f'{NER_DIR}/{d}', exist_ok=True)

print(f'NER dir: {NER_DIR}')
print(f'Code dir: {CODE_DIR}')
````

## 2. NER Extraction

Extracts biomedical entities from BioMQM source sentences using `d4data/biomedical-ner-all`.

````python
ner_output = f'{NER_DIR}/results/ner/ner_output.jsonl'

cmd = [
    sys.executable, '-u', f'{CODE_DIR}/ner_extraction.py',
    '--input_path', f'{DATA_DIR}/biomqm.jsonl',
    '--output_path', ner_output
]

print('Running NER extraction...')
subprocess.run(cmd, check=True)
print(f'✓ NER extraction → {ner_output}')
````

## 3. QG — Entity-Aware Question Generation

Generates questions specific to each extracted entity using `qg_entity_aware.py`.

````python
qg_output = f'{NER_DIR}/results/qg/qg_entity_aware.jsonl'

cmd = [
    sys.executable, '-u', f'{CODE_DIR}/qg_entity_aware.py',
    '--input_path', ner_output,
    '--output_path', qg_output
]

print('Running entity-aware QG...')
subprocess.run(cmd, check=True)
print(f'✓ QG → {qg_output}')
````

## 4. QA Source — Entity-Aware Answers on Source Sentences

````python
qa_src_output = f'{NER_DIR}/results/qa/source/qa_source.jsonl'

cmd = [
    sys.executable, '-u', f'{CODE_DIR}/qa_entity.py',
    '--mode', 'source',
    '--input_path', qg_output,
    '--output_path', qa_src_output
]

print('Running QA Source...')
subprocess.run(cmd, check=True)
print(f'✓ QA Source → {qa_src_output}')
````

## 5. QA BT — Entity-Aware Answers on Back-Translations (all languages)

````python
for lang in LANGUAGES:
    qa_bt_output = f'{NER_DIR}/results/qa/bt/qa_bt_{lang}.jsonl'
    bt_file = f'{DATA_DIR}/backtranslation/{lang}.jsonl'

    cmd = [
        sys.executable, '-u', f'{CODE_DIR}/qa_entity.py',
        '--mode', 'bt',
        '--lang', lang,
        '--input_path', qg_output,
        '--bt_path', bt_file,
        '--output_path', qa_bt_output
    ]

    print(f'Running QA BT [{lang}]...')
    subprocess.run(cmd, check=True)
    print(f'✓ QA BT [{lang}] → {qa_bt_output}')

print('\n✓ All QA complete!')
````

## 6. Mapping

````python
cmd = [
    sys.executable, '-u', f'{CODE_DIR}/mapping.py',
    '--base_dir', NER_DIR
]
print('Running mapping...')
subprocess.run(cmd, check=True)
print('✓ Mapping complete!')
````

## 7. String Comparison

````python
cmd = [
    sys.executable, '-u', f'{CODE_DIR}/string_comparison.py',
    '--base_dir', NER_DIR
]
print('Running string comparison...')
subprocess.run(cmd, check=True)
print('✓ String comparison complete!')
````

## 8. SBERT Evaluation

````python
cmd = [
    sys.executable, '-u', f'{CODE_DIR}/sbert.py',
    '--base_dir', NER_DIR
]
print('Running SBERT evaluation...')
subprocess.run(cmd, check=True)
print('✓ SBERT evaluation complete!')
````
